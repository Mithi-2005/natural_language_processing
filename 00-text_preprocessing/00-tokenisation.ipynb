### Tokenisation

```python
corpus="""Hello, Iam Mithilesh. I am very much interested in learning NLP. 
Hurray! I already complted 2 videos.It's is very Important.
"""
```

```python
print(corpus)
```

```python
import nltk
nltk.download('punkt',quiet=True)
nltk.download('punkt_tab',quiet=True)
```

```python
### Tokenisation
### para->sentences

from nltk.tokenize import  sent_tokenize ### TO tokenize to sentence
documents=sent_tokenize(corpus)
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
from nltk.tokenize import word_tokenize ### TO tokenize to word
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize ###Punctuation tokenize , Treats punctuations as seperate words like ["'","s"] before it is ["'s"]
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer

tokeniser=TreebankWordTokenizer()
tokeniser.tokenize(corpus) ## Full stop(.) not treated as another word except for last word of complete para
```
