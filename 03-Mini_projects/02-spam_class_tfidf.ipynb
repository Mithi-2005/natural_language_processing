### Spam Classification

### Flow : 
1) Preprocessing
2) Train test split
3) Vectorization
4) Training

### Preprocessing

```python
import pandas as pd

messages=pd.read_csv('../datasets/SMSSpamCollection.txt',sep='\t',names=['lables','message'])
```

```python
messages.head()
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps=PorterStemmer()
stopwrds=stopwords.words('english')
# stopwrds
```

```python
corpus=[]
for i in range(len(messages)):
    review=re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[ps.stem(word) for word in review if word not in stopwrds]
    review=" ".join(review)
    corpus.append(review)
corpus
```

### Train test split

```python
y=pd.get_dummies(messages['lables'],dtype=int)
y=y.iloc[:,1].values
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(corpus,y,train_size=0.80)
```

```python
(x_train,y_train)
```

### Vectorisation using TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfi=TfidfVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
x_train=tfi.fit_transform(x_train).toarray()
x_test=tfi.transform(x_test).toarray()
```

```python
x_train
```

```python
x_test
```

```python
tfi.vocabulary_
```

### Model Training

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_model=MultinomialNB()
```

```python
spam_model.fit(x_train,y_train)
```

```python
y_pred=spam_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report

accuracy_score(y_test,y_pred)*100
```

```python
print(classification_report(y_test,y_pred))
```
