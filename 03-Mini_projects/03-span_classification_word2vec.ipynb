```python
import pandas as pd
messages=pd.read_csv('..\datasets\SMSSpamCollection.txt',names=['labels','message'],sep='\t')
messages.head()
```

```python
from nltk.stem import  WordNetLemmatizer
from nltk.corpus import stopwords
import re
stopwrds=stopwords.words('english')
lemmatize=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(len(messages)):
    review=re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[lemmatize.lemmatize(word) for word in review]
    review=" ".join(review)
    corpus.append(review)
corpus
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
words=[]
for sent in corpus:
    sent_token=sent_tokenize(sent)
    for word in sent_token:
        words.append(simple_preprocess(word))
```

```python
words
```

```python
### Lets train word2vec from scratch
w2v_model=Word2Vec(words,epochs=10)
```

```python
import numpy as np
```

```python
def avg_w2v(doc):
    words = [word for word in doc if word in w2v_model.wv.index_to_key]
    if (len(words))==0:
        return np.zeros(w2v_model.vector_size)
    return np.mean(w2v_model.wv[words],axis=0)
```

```python
import tqdm
from tqdm import tqdm
```

```python
x=[]

for i in tqdm(range(len(words))):
    x.append(avg_w2v(words[i]))
```
